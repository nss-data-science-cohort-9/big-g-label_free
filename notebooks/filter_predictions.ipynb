```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
training_df = pd.read_csv(filepath_or_buffer='../data/training_faults_diagnostics.csv', low_memory=False)
```

```python
training_df['EventTimeStamp'] = pd.to_datetime(training_df['EventTimeStamp'])
```

```python
training_df['event_date'] = training_df['EventTimeStamp'].dt.date
```

```python
training_df[training_df['Derate_Target']==1].groupby(['EquipmentID', 'event_date'])['Derate_Target'].count().count()
```
